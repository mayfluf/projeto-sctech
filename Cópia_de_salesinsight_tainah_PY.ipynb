1. GERAÇÃO DO DATASET SINTÉTICO

`````python

def gerar_dataset_vendas(n_registros: int = 200, seed: int = 42) -> pd.DataFrame:
    """
    Gera um dataset sintético de vendas com dados intencionalmente sujos.

    Parâmetros:
        n_registros (int): Número de registros a serem gerados.
        seed (int): Semente para reprodutibilidade.

    Retorna:
        pd.DataFrame: DataFrame contendo os dados brutos de vendas.
    """
    # Define sementes para os geradores aleatórios (Python e NumPy)
    random.seed(seed)
    np.random.seed(seed)

    # Listas de valores possíveis para cada campo categórico
    produtos = ["Notebook", "Smartphone", "Tablet", "Monitor", "Teclado", "Mouse", "Headset"]
    # Dicionário que mapeia cada produto à sua categoria
    categorias = {
        "Notebook": "Computadores",
        "Smartphone": "Celulares",
        "Tablet": "Celulares",
        "Monitor": "Computadores",
        "Teclado": "Periféricos",
        "Mouse": "Periféricos",
        "Headset": "Periféricos"
    }
    regioes = ["Sudeste", "Sul", "Nordeste", "Centro-Oeste", "Norte"]
    # Gera 50 clientes com IDs formatados: Cliente_001, Cliente_002, ...
    clientes = [f"Cliente_{i:03d}" for i in range(1, 51)]

    # Data inicial: 1º de janeiro de 2024
    data_inicio = datetime(2024, 1, 1)

    dados = []  # Lista que armazenará os dicionários (cada um é uma linha/venda)

    for i in range(n_registros):
        # Escolhe aleatoriamente um produto
        produto = random.choice(produtos)

        # Quantidade entre 1 e 10 (pode se tornar None posteriormente com 5% de chance)
        quantidade = random.randint(1, 10)

        # Preço base do produto (valores médios de mercado)
        preco_base = {
            "Notebook": 3500,
            "Smartphone": 2200,
            "Tablet": 1800,
            "Monitor": 1200,
            "Teclado": 250,
            "Mouse": 120,
            "Headset": 350
        }[produto]

        # Aplica uma variação aleatória de -15% a +15% ao preço base
        preco = round(preco_base * random.uniform(0.85, 1.15), 2)

        # Gera uma data aleatória dentro do ano de 2024 (0 a 364 dias após 1º jan)
        data = data_inicio + timedelta(days=random.randint(0, 364))

        # ---------- INSERÇÃO INTENCIONAL DE SUJEIRA (dados inconsistentes) ----------
        # 5% de chance de a quantidade ser None (valor nulo)
        if random.random() < 0.05:
            quantidade = None

        # 4% de chance de o preço ser None
        if random.random() < 0.04:
            preco = None

        # 3% de chance de o produto conter espaços extras à esquerda
        if random.random() < 0.03:
            produto = "  " + produto

        # 2% de chance de a data ser uma string inválida
        data_str = data.strftime("%Y-%m-%d") if random.random() > 0.02 else "DATA INVÁLIDA"

        # Monta o dicionário representando a venda
        dados.append({
            "id_venda": i + 1,                     # Identificador único da venda
            "data_venda": data_str,                # Data (pode estar suja)
            "cliente": random.choice(clientes),    # Cliente sorteado
            "produto": produto,                    # Nome do produto (pode ter espaços extras)
            "categoria": categorias.get(produto.strip(), "Outros"),  # Categoria baseada no produto (strip para limpeza)
            "regiao": random.choice(regioes),      # Região da venda
            "quantidade": quantidade,              # Quantidade (pode ser None)
            "preco_unitario": preco                # Preço unitário (pode ser None)
        })

    # Converte a lista de dicionários em DataFrame e retorna
    df = pd.DataFrame(dados)
    logger.info(f"Dataset gerado com {len(df)} registros.")
    return df
`````

2. FUNÇÕES DE LIMPEZA DE DADOS


`````python
def limpar_dados(df: pd.DataFrame) -> tuple[pd.DataFrame, dict]:
    """
    Limpa e trata o DataFrame de vendas.

    Operações realizadas:
    - Remove espaços extras em colunas de texto.
    - Converte datas e remove linhas com datas inválidas.
    - Remove linhas com quantidade ou preco_unitario nulos.
    - Garante tipos numéricos corretos.
    - Aplica expressão regular para validar formato do cliente.

    Parâmetros:
        df (pd.DataFrame): DataFrame bruto (possivelmente sujo).

    Retorna:
        tuple: (DataFrame limpo, dicionário com relatório de limpeza)
    """
    n_inicial = len(df)
    relatorio = {}

    # Cria uma cópia para não modificar o original
    df_clean = df.copy()

    # ------------------------------------------------------------------------
    # 1. Remover espaços extras em todas as colunas de texto (strings)
    # ------------------------------------------------------------------------
    # Seleciona as colunas cujo tipo é 'object' (geralmente strings)
    colunas_texto = df_clean.select_dtypes(include="object").columns
    for col in colunas_texto:
        # strip() remove espaços no início e fim; também lida com valores None/NaN
        df_clean[col] = df_clean[col].str.strip()

    # ------------------------------------------------------------------------
    # 2. Converter coluna de data para datetime e remover inválidas (RF13 - regex indireto)
    #    Usamos pandas.to_datetime com errors='coerce' que transforma valores inválidos em NaT.
    #    Em seguida, removemos as linhas com NaT.
    # ------------------------------------------------------------------------
    df_clean["data_venda"] = pd.to_datetime(df_clean["data_venda"], errors="coerce")
    n_datas_invalidas = df_clean["data_venda"].isnull().sum()
    df_clean = df_clean.dropna(subset=["data_venda"])
    relatorio["datas_invalidas_removidas"] = n_datas_invalidas

    # ------------------------------------------------------------------------
    # 3. Remover linhas onde quantidade ou preco_unitario são nulos (NaN)
    # ------------------------------------------------------------------------
    n_antes = len(df_clean)
    df_clean = df_clean.dropna(subset=["quantidade", "preco_unitario"])
    relatorio["linhas_nulas_removidas"] = n_antes - len(df_clean)

    # ------------------------------------------------------------------------
    # 4. Garantir que os tipos numéricos estejam corretos
    # ------------------------------------------------------------------------
    df_clean["quantidade"] = df_clean["quantidade"].astype(int)
    df_clean["preco_unitario"] = df_clean["preco_unitario"].astype(float)

    # ------------------------------------------------------------------------
    # 5. Uso de expressão regular (RF13) para limpeza e validação de clientes
    # ------------------------------------------------------------------------
    def validar_cliente(nome_cliente: str) -> tuple[str, bool]:
        """
        Aplica regex para limpar e validar o nome do cliente.
        Espera-se o padrão: Cliente_ seguido de exatamente três dígitos.
        """
        # Remove qualquer caractere que não seja letra, dígito, underline ou espaço
        limpo = re.sub(r"[^a-zA-Z0-9_ ]", "", nome_cliente).strip()
        # Verifica se o nome limpo corresponde ao padrão Cliente_XXX
        valido = bool(re.fullmatch(r"Cliente_\d{3}", limpo))
        return limpo, valido

    # Aplica a função em cada linha da coluna 'cliente'
    resultado = df_clean["cliente"].apply(validar_cliente)
    df_clean["cliente_limpo"] = resultado.apply(lambda x: x[0])   # Primeiro elemento: nome limpo
    df_clean["cliente_valido"] = resultado.apply(lambda x: x[1])  # Segundo elemento: booleano

    n_invalidos = (~df_clean["cliente_valido"]).sum()
    relatorio["clientes_formato_invalido"] = n_invalidos

    # Opcional: podemos substituir a coluna original pela limpa (mas manteremos ambas para análise)
    df_clean["cliente"] = df_clean["cliente_limpo"]

    # ------------------------------------------------------------------------
    # 6. Registrar estatísticas finais
    # ------------------------------------------------------------------------
    n_final = len(df_clean)
    relatorio["registros_iniciais"] = n_inicial
    relatorio["registros_finais"] = n_final
    relatorio["registros_removidos_total"] = n_inicial - n_final

    # Exibe o relatório formatado no console via logging
    logger.info("=== RELATÓRIO DE LIMPEZA ===")
    for chave, valor in relatorio.items():
        logger.info(f"  {chave}: {valor}")

    return df_clean, relatorio
`````

3. CRIAÇÃO DE COLUNAS DERIVADAS

`````python
def criar_colunas_derivadas(df: pd.DataFrame) -> pd.DataFrame:
    """
    Cria novas colunas calculadas a partir dos dados existentes.

    Colunas adicionadas:
    - receita_total: quantidade * preco_unitario
    - mes: número do mês (1 a 12)
    - mes_nome: nome do mês em inglês (ou português dependendo da localidade)
    - trimestre: trimestre no formato Q1, Q2, Q3, Q4
    - ano: ano extraído da data
    - faixa_receita_item: classificação da receita (Baixo/Médio/Alto Valor) usando np.select

    Parâmetros:
        df (pd.DataFrame): DataFrame limpo.

    Retorna:
        pd.DataFrame: DataFrame com as novas colunas.
    """
    # Cria uma cópia para evitar warnings de modificação
    df_transform = df.copy()

    # Receita total por linha (produto escalar)
    df_transform["receita_total"] = df_transform["quantidade"] * df_transform["preco_unitario"]

    # Extrações de data: .dt acessa componentes datetime
    df_transform["mes"] = df_transform["data_venda"].dt.month
    df_transform["mes_nome"] = df_transform["data_venda"].dt.strftime("%B")  # Nome completo do mês
    df_transform["trimestre"] = df_transform["data_venda"].dt.quarter.apply(lambda q: f"Q{q}")  # Ex: Q1
    df_transform["ano"] = df_transform["data_venda"].dt.year

    # Classificação da receita por item usando numpy.select (vetorizado e eficiente)
    condicoes = [
        df_transform["receita_total"] < 500,                         # Baixo
        (df_transform["receita_total"] >= 500) & (df_transform["receita_total"] < 5000),  # Médio
        df_transform["receita_total"] >= 5000                       # Alto
    ]
    classificacoes = ["Baixo Valor", "Médio Valor", "Alto Valor"]
    df_transform["faixa_receita_item"] = np.select(condicoes, classificacoes, default="Não Classificado")

    logger.info("Colunas derivadas criadas com sucesso.")
    logger.debug(f"Amostra das novas colunas:\n{df_transform[['receita_total','mes','trimestre','faixa_receita_item']].head()}")
    return df_transform
`````

4. MÉTRICAS AGREGADAS

`````python
def calcular_metricas(df: pd.DataFrame) -> dict:
    """
    Calcula métricas agregadas usando groupby.

    Métricas:
    - Receita total e quantidade vendida por mês.
    - Top 5 produtos por receita.
    - Receita total por categoria.
    - Receita total e ticket médio por região.

    Parâmetros:
        df (pd.DataFrame): DataFrame com colunas derivadas.

    Retorna:
        dict: Dicionário contendo cada tabela de métricas.
    """
    metricas = {}

    # 1. Receita e quantidade por mês
    por_mes = df.groupby("mes").agg(
        receita_total=("receita_total", "sum"),
        quantidade=("quantidade", "sum"),
        n_vendas=("id_venda", "count")
    ).reset_index().sort_values("mes")
    metricas["por_mes"] = por_mes

    # 2. Top 5 produtos por receita
    top_produtos = df.groupby("produto")["receita_total"].sum() \
                     .sort_values(ascending=False).head(5).reset_index()
    metricas["top_produtos"] = top_produtos

    # 3. Receita por categoria
    por_categoria = df.groupby("categoria")["receita_total"].sum().reset_index()
    metricas["por_categoria"] = por_categoria

    # 4. Receita e ticket médio por região
    por_regiao = df.groupby("regiao").agg(
        receita_total=("receita_total", "sum"),
        media_ticket=("receita_total", "mean")  # ticket médio por venda naquela região
    ).reset_index().sort_values("receita_total", ascending=False)
    metricas["por_regiao"] = por_regiao

    # Exibição formatada via logging
    logger.info("=== MÉTRICAS AGREGADAS ===")
    for nome, tabela in metricas.items():
        logger.info(f"\n--- {nome.upper().replace('_', ' ')} ---")
        # Converte para string sem índice para exibição limpa
        logger.info("\n" + tabela.to_string(index=False))

    return metricas
`````

5. SEGMENTAÇÃO DE CLIENTES - com lambda

`````python
def segmentar_clientes(df: pd.DataFrame) -> pd.DataFrame:
    """
    Segmenta clientes baseado no total gasto.
    Critérios:
        - Bronze: gasto < R$ 5.000
        - Prata: R$ 5.000 <= gasto <= R$ 15.000
        - Ouro: gasto > R$ 15.000

    Utiliza groupby e apply com função lambda para classificação.

    Parâmetros:
        df (pd.DataFrame): DataFrame contendo coluna 'cliente' e 'receita_total'.

    Retorna:
        pd.DataFrame: DataFrame com clientes, total_gasto e segmento.
    """
    # Agrupa por cliente e soma a receita total
    clientes = df.groupby("cliente")["receita_total"].sum().reset_index()
    clientes.columns = ["cliente", "total_gasto"]

    # Classifica usando apply com lambda (demonstração de função lambda)
    clientes["segmento"] = clientes["total_gasto"].apply(
        lambda gasto: "Ouro" if gasto > 15000
                      else ("Prata" if gasto >= 5000 else "Bronze")
    )

    # Ordena do maior gasto para o menor
    clientes = clientes.sort_values("total_gasto", ascending=False)

    logger.info("=== SEGMENTAÇÃO DE CLIENTES ===")
    logger.info(f"\nTop 10 clientes:\n{clientes.head(10).to_string(index=False)}")
    logger.info(f"\nDistribuição dos segmentos:\n{clientes['segmento'].value_counts().to_string()}")

    return clientes
`````

6. ESTATÍSTICAS COM NUMPY

`````python
def calcular_estatisticas_numpy(df: pd.DataFrame) -> tuple[dict, np.ndarray, np.ndarray]:
    """
    Usa NumPy para calcular estatísticas sobre as receitas.
    Demonstra:
        - Conversão para array NumPy
        - Operações vetorizadas (média, mediana, desvio, percentis)
        - Broadcasting (normalização)
        - Filtro vetorizado (vendas acima da média)

    Parâmetros:
        df (pd.DataFrame): DataFrame com coluna 'receita_total'.

    Retorna:
        tuple: (Dicionário com as estatísticas calculadas, Array NumPy de receitas normalizadas, Array NumPy de vendas acima da média).
    """
    # Converte a coluna 'receita_total' para um array NumPy (1D)
    receitas = df["receita_total"].to_numpy()

    # Estatísticas básicas usando funções NumPy
    media = np.mean(receitas)
    mediana = np.median(receitas)
    desvio_padrao = np.std(receitas)
    total = np.sum(receitas)
    p25 = np.percentile(receitas, 25)   # Primeiro quartil
    p75 = np.percentile(receitas, 75)   # Terceiro quartil

    logger.info("=== ESTATÍSTICAS COM NUMPY ===")
    logger.info(f"  Receita média por venda:    R$ {media:.2f}")
    logger.info(f"  Receita mediana por venda:  R$ {mediana:.2f}")
    logger.info(f"  Desvio padrão:              R$ {desvio_padrao:.2f}")
    logger.info(f"  Receita total:              R$ {total:,.2f}")
    logger.info(f"  Percentil 25 (Q1):          R$ {p25:.2f}")
    logger.info(f"  Percentil 75 (Q3):          R$ {p75:.2f}")

    # Broadcasting: normalização min-max (coloca os valores entre 0 e 1)
    # (receitas - min) / (max - min)
    min_receita = receitas.min()
    max_receita = receitas.max()
    if max_receita > min_receita:
        receitas_normalizadas = (receitas - min_receita) / (max_receita - min_receita)
    else:
        receitas_normalizadas = np.zeros_like(receitas)
    logger.info(f"\n  Receitas normalizadas (primeiros 5): {receitas_normalizadas[:5].round(4)}")

    # Operação vetorizada: identificar vendas acima da média sem usar loops explícitos
    acima_da_media = receitas[receitas > media]   # Filtro booleano
    logger.info(f"\n  Vendas acima da média: {len(acima_da_media)} de {len(receitas)} ({len(acima_da_media)/len(receitas):.1%})")

    stats_dict = {
        "media": media,
        "mediana": mediana,
        "desvio_padrao": desvio_padrao,
        "total": total,
        "percentil_25": p25,
        "percentil_75": p75
    }

    return stats_dict, receitas_normalizadas, acima_da_media
`````

8. FUNÇÃO DE ORDEM SUPERIOR (HIGHER-ORDER FUNCTION)

`````python
def processar_coluna(df: pd.DataFrame, coluna: str, funcao_transformacao) -> pd.DataFrame:
    """
    Aplica uma função de transformação a uma coluna do DataFrame.
    Esta é uma função de ordem superior (higher-order function) pois recebe outra função como argumento.

    Parâmetros:
        df (pd.DataFrame): DataFrame de entrada.
        coluna (str): Nome da coluna a ser transformada.
        funcao_transformacao: Função que recebe um valor e retorna o valor transformado.

    Retorna:
        pd.DataFrame: DataFrame com uma nova coluna '{coluna}_transformado'.
    """
    df_result = df.copy()
    nova_coluna = f"{coluna}_transformado"
    df_result[nova_coluna] = df_result[coluna].apply(funcao_transformacao)
    logger.info(f"Coluna '{nova_coluna}' criada aplicando função de ordem superior.")
    return df_result
`````

`````python
def exportar_resultados(metricas: dict, clientes: pd.DataFrame, stats_numpy: dict, output_dir: str = "outputs") -> None:
    """
    Exporta os resultados em formatos CSV e JSON.

    Parâmetros:
        metricas (dict): Dicionário com DataFrames de métricas (por_mes, top_produtos, etc.)
        clientes (pd.DataFrame): DataFrame com segmentação de clientes.
        stats_numpy (dict): Estatísticas calculadas com NumPy.
        output_dir (str): Diretório onde os arquivos serão salvos.
    """
    os.makedirs(output_dir, exist_ok=True)

    # 1. Exportar métricas por mês para CSV
    caminho_csv1 = os.path.join(output_dir, "metricas_por_mes.csv")
    metricas["por_mes"].to_csv(caminho_csv1, index=False, encoding="utf-8-sig")
    logger.info(f"CSV exportado: {caminho_csv1}")

    # 2. Exportar segmentação de clientes para CSV
    caminho_csv2 = os.path.join(output_dir, "segmentacao_clientes.csv")
    clientes.to_csv(caminho_csv2, index=False, encoding="utf-8-sig")
    logger.info(f"CSV exportado: {caminho_csv2}")

    # 3. Exportar estatísticas gerais para JSON
    caminho_json = os.path.join(output_dir, "estatisticas_gerais.json")
    # Converte valores numpy para float nativo (serializável)
    stats_serializaveis = {k: float(v) if isinstance(v, (np.float32, np.float64)) else v for k, v in stats_numpy.items()}
    with open(caminho_json, "w", encoding="utf-8") as f:
        json.dump(stats_serializaveis, f, indent=4, ensure_ascii=False)
    logger.info(f"JSON exportado: {caminho_json}")

    # 4. (Opcional) Ler o JSON exportado e exibir uma amostra para verificação
    with open(caminho_json, "r", encoding="utf-8") as f:
        dados_lidos = json.load(f)
    logger.debug(f"Conteúdo do JSON lido: {json.dumps(dados_lidos, indent=2)}")
`````

10. CLASSES - Herança

`````python
class AnalisadorDeVendas:
    """
    Classe base responsável por encapsular o pipeline de análise de vendas.
    Mantém o estado do DataFrame e resultados intermediários.
    """

    def __init__(self, caminho_arquivo: str):
        """
        Construtor da classe.

        Parâmetros:
            caminho_arquivo (str): Caminho para o arquivo CSV contendo os dados de vendas.
        """
        self.caminho_arquivo = caminho_arquivo
        self.df_bruto = None          # DataFrame antes da limpeza
        self.df_limpo = None          # DataFrame após limpeza e transformações
        self.metricas = {}            # Dicionário com métricas agregadas
        self.clientes = None          # DataFrame com segmentação de clientes
        self.relatorio_limpeza = {}   # Relatório da etapa de limpeza

    def carregar(self) -> 'AnalisadorDeVendas':
        """Lê o arquivo CSV e armazena o DataFrame bruto."""
        self.df_bruto = pd.read_csv(self.caminho_arquivo)
        logger.info(f"Arquivo carregado: {self.caminho_arquivo}")
        logger.info(f"  Registros carregados: {len(self.df_bruto)}")
        return self   # Retorna a própria instância para permitir encadeamento (fluent interface)

    def limpar(self) -> 'AnalisadorDeVendas':
        """Limpa os dados e armazena o DataFrame tratado."""
        self.df_limpo, self.relatorio_limpeza = limpar_dados(self.df_bruto.copy())
        return self

    def transformar(self) -> 'AnalisadorDeVendas':
        """Aplica transformações e cria colunas derivadas."""
        self.df_limpo = criar_colunas_derivadas(self.df_limpo)
        return self

    def analisar(self) -> 'AnalisadorDeVendas':
        """Calcula métricas e segmentações."""
        self.metricas = calcular_metricas(self.df_limpo)
        self.clientes = segmentar_clientes(self.df_limpo)
        return self

    def visualizar(self) -> 'AnalisadorDeVendas':
        """Gera e exporta os gráficos."""
        gerar_visualizacoes(self.df_limpo, self.metricas)
        return self

    def exportar_relatorio(self, caminho: str = "outputs/relatorio_resumo.csv") -> 'AnalisadorDeVendas':
        """Exporta o relatório de métricas por mês em CSV."""
        os.makedirs("outputs", exist_ok=True)
        self.metricas["por_mes"].to_csv(caminho, index=False)
        logger.info(f"Relatório resumo exportado: {caminho}")
        return self

    def resumo(self) -> None:
        """Exibe um resumo executivo do pipeline."""
        logger.info("\n" + "="*50)
        logger.info("       RESUMO EXECUTIVO – SALESINSIGHT PY")
        logger.info("="*50)
        logger.info(f"  Arquivo analisado:      {self.caminho_arquivo}")
        logger.info(f"  Registros brutos:       {self.relatorio_limpeza.get('registros_iniciais', 'N/A')}")
        logger.info(f"  Registros limpos:       {self.relatorio_limpeza.get('registros_finais', 'N/A')}")
        receita = self.df_limpo["receita_total"].sum() if self.df_limpo is not None else 0
        logger.info(f"  Receita total anual:    R$ {receita:,.2f}")
        if self.clientes is not None and len(self.clientes) > 0:
            top_cliente = self.clientes.iloc[0]
            logger.info(f"  Cliente top:            {top_cliente['cliente']} (R$ {top_cliente['total_gasto']:,.2f})")
        logger.info("="*50)


class AnalisadorComProjecao(AnalisadorDeVendas):
    """
    Classe derivada que estende AnalisadorDeVendas com funcionalidade de projeção de tendência.
    Demonstra herança e uso de super().
    """

    def __init__(self, caminho_arquivo: str, meses_projecao: int = 3):
        """Inicializa a classe pai e adiciona atributos específicos."""
        super().__init__(caminho_arquivo)
        self.meses_projecao = meses_projecao
        self.projecoes = []          # Lista de dicionários com meses projetados e valores

    def projetar_tendencia(self) -> 'AnalisadorComProjecao':
        """
        Projeta a receita dos próximos meses usando regressão linear simples.
        Essa abordagem é mais robusta que a média móvel e demonstra uso avançado de NumPy.
        """
        if not self.metricas or "por_mes" not in self.metricas:
            logger.warning("Métricas não encontradas. Execute .analisar() antes de projetar.")
            return self

        # Obtém a tabela de receita por mês e ordena
        por_mes = self.metricas["por_mes"].sort_values("mes")
        receitas_historicas = por_mes["receita_total"].to_numpy()
        meses_historicos = np.arange(1, len(receitas_historicas) + 1)

        # Ajusta uma reta de regressão linear (grau 1) usando polyfit
        # Retorna (coeficiente_angular, coeficiente_linear)
        coeficientes = np.polyfit(meses_historicos, receitas_historicas, 1)
        tendencia = np.poly1d(coeficientes)   # Cria uma função polinomial

        ultimo_mes = int(por_mes["mes"].max())
        self.projecoes = []

        logger.info("=== PROJEÇÃO DE TENDÊNCIA (Regressão Linear) ===")
        logger.info(f"  Coeficiente angular (crescimento mensal): R$ {coeficientes[0]:.2f}")
        for i in range(1, self.meses_projecao + 1):
            mes_projetado = ultimo_mes + i
            receita_projetada = tendencia(mes_projetado)
            # Garante que a projeção não seja negativa
            receita_projetada = max(0, receita_projetada)
            self.projecoes.append({
                "mes": mes_projetado,
                "receita_projetada": round(receita_projetada, 2)
            })
            logger.info(f"  Mês {mes_projetado:02d} (projeção): R$ {receita_projetada:,.2f}")

        return self

    def exibir_projecao_detalhada(self) -> None:
        """Exibe as projeções calculadas de forma detalhada."""
        if not self.projecoes:
            logger.warning("Nenhuma projeção disponível. Rode .projetar_tendencia() primeiro.")
            return
        logger.info("\n=== DETALHAMENTO DAS PROJEÇÕES ===")
        for p in self.projecoes:
            logger.info(f"  Mês {p['mes']:02d}: R$ {p['receita_projetada']:,.2f}")

`````

11. PIPELINE PRINCIPAL

`````python
def main():
    """
    Função principal que orquestra todo o pipeline:
    - Gera o dataset se não existir.
    - Instancia a classe com projeção.
    - Executa as etapas: carregar, limpar, transformar, analisar, projetar, visualizar, exportar.
    - Aplica funções adicionais (higher-order function, regex já integrado).
    - Exporta resultados finais.
    """
    logger.info("="*60)
    logger.info("   SALESINSIGHT PY – Pipeline de Análise de Dados de Vendas")
    logger.info("="*60)

    # Verifica se o arquivo CSV já existe. Se não, gera um novo dataset.
    if not os.path.exists("vendas.csv"):
        logger.info("Dataset 'vendas.csv' não encontrado. Gerando dataset sintético...")
        df_gerado = gerar_dataset_vendas(n_registros=200, seed=42)
        df_gerado.to_csv("vendas.csv", index=False)
        logger.info("Dataset gerado e salvo como 'vendas.csv'.")

    # Instancia o analisador com projeção (herança)
    analisador = AnalisadorComProjecao("vendas.csv", meses_projecao=3)

    # Executa o pipeline encadeado (fluent interface)
    (analisador
        .carregar()
        .limpar()
        .transformar()
        .analisar()
        .projetar_tendencia()
        .visualizar()
        .exportar_relatorio()
    )

    # Demonstração da função de ordem superior (RF11) - aplica uma transformação lambda
    logger.info("\n--- Demonstração de Higher-Order Function (processar_coluna) ---")
    # Aplica uma função lambda que arredonda a receita total para milhares
    analisador.df_limpo = processar_coluna(analisador.df_limpo, "receita_total", lambda x: round(x / 1000, 2))
    # Exibe as primeiras linhas da nova coluna criada
    logger.info(f"Amostra da coluna 'receita_total_transformado':\n{analisador.df_limpo[['receita_total', 'receita_total_transformado']].head().to_string()}")

    # Calcula estatísticas NumPy e exporta resultados adicionais
    logger.info("\n--- Estatísticas com NumPy ---")
    stats = calcular_estatisticas_numpy(analisador.df_limpo)
    exportar_resultados(analisador.metricas, analisador.clientes, stats)

    # Exibe resumo executivo
    analisador.resumo()
    analisador.exibir_projecao_detalhada()

    logger.info("\n[CONCLUÍDO] Pipeline finalizado com sucesso! Verifique a pasta 'outputs/'.")
`````

`````python
# PONTO DE ENTRADA
if __name__ == "__main__":
    main()
`````
